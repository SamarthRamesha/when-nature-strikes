# WHEN NATURE STRIKES
## Exploratory Data Analysis of Natural Disasters

### Project Objective
To analyse global natural disaster data from EM-DAT
and investigate whether the most frequent disaster
types are also the most destructive.

### Analysis Areas
1. Disaster frequency by type
2. Human impact analysis
3. Economic damage analysis
4. Yearly and decadal trends
5. Frequency versus destructiveness

### Dataset
EM-DAT Natural Disaster Records (1900–2026)

### Data Source
EM-DAT - CRED / UCLouvain

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Visualization settings
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

print("All libraries imported successfully!")
```

```python
# Load processed datasets

events = pd.read_csv(
    "../data/processed/natural_disaster_events.csv",
    encoding="utf-8-sig"
)

type_summary = pd.read_csv(
    "../data/processed/natural_disaster_type_summary.csv",
    encoding="utf-8-sig"
)

print("Datasets loaded successfully!")
```

```python
print("EVENT-LEVEL DATASET")
print("Rows:", events.shape[0])
print("Columns:", events.shape[1])

print("\nDISASTER TYPE SUMMARY")
print("Rows:", type_summary.shape[0])
print("Columns:", type_summary.shape[1])
```

```python
display(events.head())
display(type_summary.head())
```

```python
print("EVENT DATASET COLUMNS:")
for column in events.columns:
    print(column)

print("\nTYPE SUMMARY COLUMNS:")
for column in type_summary.columns:
    print(column)
```

```python
#CHECK DUPLICATE EVENT IDs
```

```python
print("Total records:", len(events))

print(
    "Unique Event IDs:",
    events["Event ID"].nunique()
)

print(
    "Duplicate Event IDs:",
    events["Event ID"].duplicated().sum()
)
```

```python
#CHECK MISSING VALUES
```

```python
missing_values = events.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (
        missing_values / len(events) * 100
    ).round(2)
})

display(
    missing_summary.sort_values(
        by="Missing Percentage",
        ascending=False
    )
)
```
